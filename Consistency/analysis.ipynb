```python
# List of the model prediction column names
model_result_columns = [
    'llama_temp0', 
    'llama_temp1',  
    'ds_temp0', 
    'ds_temp1',
    'gemini_label',
    'chatgpt_label'
]
```

```python
# Human labelled column
actual_label = data_one['label']

# Collect performance metrics for each model
results = []

for model_name in model_result_columns:
    predicted_label = data_one[model_name]
    
    # Calculate performance accuracy
    accuracy = accuracy_score(actual_label, predicted_label)
    
    # Evaluation metrics
    precision = precision_score(actual_label, predicted_label, average='macro', zero_division=0)
    recall = recall_score(actual_label, predicted_label, average='macro', zero_division=0)
    F1_score = f1_score(actual_label, predicted_label, average='macro', zero_division=0)
    

    results.append([
        model_name,
        f"{accuracy:.4f}",
        f"{precision:.4f}",
        f"{recall:.4f}",
        f"{F1_score:.4f}",
    ])

# Table headers
Table_column_names = [
    "Model", 
    "Accuracy", 
    "Precision", 
    "Recall", 
    "F1-Score", 
]

# Print formatted table using tabulate
print(tabulate(results, headers=Table_column_names, tablefmt="grid"))
```

```python
# Create a 2x3 subplot grid
figure, axes = plt.subplots(2, 3, figsize=(14, 8))
axes = axes.flatten()

# Generate Confusion Matrix heatmap for each model
for index, model in enumerate(model_result_columns):
    ConfusionMatrix = confusion_matrix(data_one["label"], data_one[model])

    sns.heatmap(
        ConfusionMatrix,
        annot=True,
        fmt="d",
        cmap="Blues",
        ax=axes[index],
        cbar=False,
        annot_kws={"size": 12, "weight": "bold"},
    )

    axes[index].set_title(
        f"Confusion Matrix: {model}", fontsize=11, fontweight="bold"
    )
    axes[index].set_xlabel("Predicted Label", fontsize=10)
    axes[index].set_ylabel("Actual Label", fontsize=10)

plt.suptitle("Confusion Matrices Across Models", fontsize=16, fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
data_labelling_results = pd.read_csv("/Users/jisha/Desktop/Sarcasm_Final/Data_Labelling/data_labeling.csv")
consistancy_test_results = pd.read_csv("/Users/jisha/Desktop/Sarcasm_Final/Consistency/consistency.csv")
```

```python
# Models to compare
model_cols = ["llama_temp0", "llama_temp1", "ds_temp0", "ds_temp1","gemini_label","chatgpt_label"]
```

```python
def interpret_kappa(k):
    if k < 0:
        return "Poor"
    elif k <= 0.20:
        return "Slight"
    elif k <= 0.40:
        return "Fair"
    elif k <= 0.60:
        return "Moderate"
    elif k <= 0.80:
        return "Substantial"
    else:
        return "Almost Perfect"
```

```python
# Tabular Repeatability Analysis
results = []

for model in model_cols:
    result_1 = data_labelling_results[model]
    result_2 = consistancy_test_results[model]

    total = len(result_1)
    matches = (result_1 == result_2).sum()
    flips = total - matches
    consistency_pct = (matches / total) * 100

    # Directional flips
    flip_0_to_1 = ((result_1 == 0) & (result_2 == 1)).sum()
    flip_1_to_0 = ((result_1 == 1) & (result_2 == 0)).sum()

    # Cohen's Kappa
    kappa = cohen_kappa_score(result_1, result_2)
    kappa_interp = interpret_kappa(kappa)

    # McNemar's Test for systematic shift
    contingency_table = pd.crosstab(result_1, result_2)

    if contingency_table.shape == (2, 2):
        mcnemar_p = mcnemar(contingency_table.values, exact=True).pvalue
        shift_bias = (
            "No Bias Shift" if mcnemar_p >= 0.05 else "Systematic Shift"
        )
    else:
        mcnemar_p = np.nan
        shift_bias = "N/A"

    results.append(
        [
            model,
            f"{matches}/{total}",
            f"{consistency_pct:.2f}%",
            flips,
            f"{flip_0_to_1} (0→1) | {flip_1_to_0} (1→0)",
            f"{kappa:.4f}",
            kappa_interp,
            f"{mcnemar_p:.4f}" if not np.isnan(mcnemar_p) else "N/A",
            shift_bias,
        ]
    )

```

```python
Table_column_names = [
    "Model Name",
    "Matches",
    "Consistency (%)",
    "Total Flips",
    "Flip Breakdown (R1→R2)",
    "Cohen's Kappa",
    "Reliability",
    "McNemar p-val",
    "Bias Shift Check",
]
print("Data Labeling  vs. Consistency  Results Summary")
print(tabulate(results, headers=Table_column_names, tablefmt="grid"))
```
